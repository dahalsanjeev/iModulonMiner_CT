### compare to original log_tpm to make sure the new script is outputting similar TPMs

```python
import pandas as pd
import numpy as np
from scipy.stats import pearsonr
```

```python
file_og = '/Users/skd6150/Documents/Softwares/modulome-workflow-main/data/raw_data/log_tpm_all_for_checking.csv' #this is just a copy of the original log_tpm.csv
file_test = '../test/nf_results/log_tpm.csv'

df_og = pd.read_csv(file_og, index_col=0)
df_test = pd.read_csv(file_test, index_col=0)
```

```python
df_test.head()
```

```python
dict_exp_corr = dict()

for col in df_test.columns.to_list():
    x = df_test[col].to_list()
    y = df_og[col].to_list()
    
    r,p = pearsonr(x, y)

    if r<0.95:
        print(f'{col} correlation is low. Check.')
    
    dict_exp_corr[col] = np.round(r,3), np.round(p,3)

print(dict_exp_corr)
```
