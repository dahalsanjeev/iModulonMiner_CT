```python
import os
```

```python
for d1 in os.listdir('./work/'):
    if d1 != '.DS_Store':
        for d2 in os.listdir(f'./work/{d1}'):
            if d2 != '.DS_Store':
                for file in os.listdir(f'./work/{d1}/{d2}', ):
                    # print(file)
                    if file == '.command.out':
                        with open(f'./work/{d1}/{d2}/{file}', 'r') as f:
                            for line in f:
                                if line.strip().startswith('Output files'):
                                    break
                                else:
                                    print(f'./work/{d1}/{d2}/{file}', line.strip())
                    
                    
```

```python
for d1 in os.listdir('./work/'):
    if d1 != '.DS_Store':
        for d2 in os.listdir(f'./work/{d1}'):
            if d2 != '.DS_Store':
                for file in os.listdir(f'./work/{d1}/{d2}'):
                    if file == '.command.err':
                        with open(f'./work/{d1}/{d2}/{file}', 'r') as f:
                            for line in f:
                                print(f'./work/{d1}/{d2}/{file}', line.strip())
                    
                    
```
