```python
import numpy as np
import pandas as pd

import re

from os import path
```

### Inputs

Enter path of QC1 passed metadata file and the metadata curated previously

```python
metadata_qc_file = path.join('.', 'output_qc', 'interim', 'metadata_qc_part1.tsv') # Enter filename for metadata QC file with only passing datasets from P1
metadata_qc_curated_file = path.join('.', 'output_qc', 'interim', 'sra_growth_metadata_ctherm_modified.xlsx') # Enter filename for metadata QC file with only passing datasets

metadata_qc_sra_curator_file = path.join('.', 'output_qc', 'interim', 'sra_growth_metadata_ctherm.csv') # Enter filename for metadata QC file with only passing datasets and created by sra_metadata_retreiver.py
```

```python
df_metadata_qc = pd.read_csv(metadata_qc_file, sep='\t') #metadata that passed the QC1
df_metadata_qc.rename(columns={'Unnamed: 0': 'SRA_Accession'}, inplace=True)
df_metadata_qc.head()
```

```python
df_metadata_curated_qc = pd.read_excel(metadata_qc_curated_file, sheet_name='curated_GOLD_for_circulation') #metadata that was curated previosuly from GOLD database
df_metadata_curated_qc.head()


# find columns that are not necessary
remove_cols = ['geo_loc_name', 'lat_lon', 'num_replicons', 'project_name', 'isol_growth_condt', 'env_broad_scale', 'ref_biomaterial', 'env_local_scale', 'collection_date', 'env_package', 'env_medium', 'investigation_type', 'cell_line', 'tissue', 'isolation_source', 'sample_type', 'sample time', 'Reactor', 'host', 'growth phase', 'isolate']
df_metadata_curated_qc_v1 = df_metadata_curated_qc.loc[:, [k for k in df_metadata_curated_qc.columns if k not in remove_cols]]
```

```python
df_metadata_merged = pd.merge(df_metadata_qc, df_metadata_curated_qc_v1, on='SRA_Accession', how='inner')
print(df_metadata_merged.shape)

df_metadata_merged.head()

```

```python
# manually remove these ids that need more curation
manual_add = ['SRX3200203',
                'SRX3200204',
                'SRX3200205',
                'SRX3200206',
                'SRX3200207',
                'SRX3200208',
                'SRX3200209',
                'SRX5209276',
                'SRX5209278']
df_metadata_merged = df_metadata_merged.loc[~df_metadata_merged['SRA_Accession'].isin(manual_add)]
```

```python
no_data_sra_id = [k for k in df_metadata_qc['SRA_Accession'].to_list() if k not in df_metadata_merged['SRA_Accession'].to_list()]


df_metadata_qc_left = df_metadata_qc.loc[df_metadata_qc['SRA_Accession'].isin(no_data_sra_id)]
df_metadata_qc_left
```

```python
df_metadata_qc_left.loc[df_metadata_qc_left.SRA_Accession=='SRX5209276']
```

```python
df_metadata_merged_v1 = pd.concat([df_metadata_merged, df_metadata_qc_left], axis=0)
df_metadata_merged_v1.head()
```

```python
# load the output data from sra_metadata_retreiver.py
df_metadata_sra_curator = pd.read_csv(metadata_qc_sra_curator_file)
```

```python
df_eh_curator = pd.read_excel('./output_qc/interim/sra_growth_metadata_ctherm_modified.xlsx', sheet_name='mapping_data_for_EH')
```

```python
df_metadata_merged_v1['proj_special'] = ''

# df = pd.DataFrame(columns=['proj', 'date_exp', 'strain'])
                  
for acc in df_metadata_qc_left['SRA_Accession'].to_list():
    if acc in df_metadata_sra_curator['accession'].to_list():
            df_row = df_metadata_sra_curator.loc[df_metadata_sra_curator.accession==acc]
            sample_name = df_row['bs__title'].values[0]

            if re.match(r'^Ruminiclostridium thermocellum \w+ transcriptomics - ', sample_name) or re.match(r'^Ruminiclostridium thermocellum DSM1313 \w+ transcriptomics - ', sample_name):
                # date_exp = re.findall(r'_\d+_\d+_\d+_', sample_name)
                # date_exp = date_exp[0].strip('_')
                # date_exp = date_exp.replace('_', '-')
                
                # proj_name = re.findall(r'LL\d+ transcriptomics', sample_name)
                # proj_name = re.sub('transcriptomics$', '', proj_name[0])

            
                # proj_name = re.findall(r'_C(I|II|III|IV)_', sample_name)
                # proj_name = re.sub('_', '', proj_name[0])
                # if proj_name=='I' or proj_name=='II':
                #      proj_name_mod = 'I-II'
                # elif proj_name=='III' or proj_name=='IV':
                #      proj_name_mod = 'III-IV'
                     

                # # # proj_name = re.findall(r'LL\d+ transcriptomics', sample_name)
                # # # proj_name = re.sub('transcriptomics$', '', proj_name[0])

                # # # if proj_name not in dict_proj_strains.keys():
                # # #     dict_proj_strains[proj_name] = list()
                
                sample_name = re.sub(r'^Ruminiclostridium thermocellum \w+ transcriptomics - ', '', sample_name)
                sample_name = re.sub(r'-v\d$', '', sample_name)
                strain_name = re.findall(r'LL\d+', sample_name)
                strain_name = strain_name[0]

                # df.loc[acc] = proj_name, date_exp, strain_name
                
                df_metadata_merged_v1.loc[df_metadata_merged_v1.SRA_Accession==acc, 'strain'] = strain_name

                # df_metadata_merged_v1.loc[df_metadata_merged_v1.SRA_Accession==acc, 'proj_special'] = 'eh_chemostat_C' + proj_name_mod
                df_metadata_merged_v1.loc[df_metadata_merged_v1.SRA_Accession==acc, 'proj_special'] = 'eh_chemostat'

                
                if sample_name.endswith('_Nlimit'):
                    df_metadata_merged_v1.loc[df_metadata_merged_v1.SRA_Accession==acc, 'treatment'] = 'N-limited'
                else:
                    df_metadata_merged_v1.loc[df_metadata_merged_v1.SRA_Accession==acc, 'treatment'] = df_row['bs__treatment'].values[0]

                df_metadata_merged_v1.loc[df_metadata_merged_v1.SRA_Accession==acc, 'batch'] = 0
                
            else:
                df_metadata_merged_v1.loc[df_metadata_merged_v1.SRA_Accession==acc, 'strain'] = df_row['bs__strain'].values[0]
                df_metadata_merged_v1.loc[df_metadata_merged_v1.SRA_Accession==acc, 'treatment'] = df_row['bs__treatment'].values[0]
            
            df_metadata_merged_v1.loc[df_metadata_merged_v1.SRA_Accession==acc, 'csource'] = df_row['bs__carbon_source'].values[0]
            df_metadata_merged_v1.loc[df_metadata_merged_v1.SRA_Accession==acc, 'time point sampled'] = df_row['bs__time_point_sampled'].values[0]
            df_metadata_merged_v1.loc[df_metadata_merged_v1.SRA_Accession==acc, 'sulfur source'] = df_row['bs__sulfur_source'].values[0]
            df_metadata_merged_v1.loc[df_metadata_merged_v1.SRA_Accession==acc, 'pH'] = df_row['bs__ph_of_chemostat'].values[0]
```

```python
# add publication data from SRA curated file
keep_cols = ['accession', 'pub__pmid', 'pub__title', 'pub__first_author', 'pub__authors', 'pub__journal', 'pub__year', 'pub__doi']

df_metadata_sra_curator_v1 = df_metadata_sra_curator[keep_cols]


df_metadata_sra_curator_v1.rename(columns={'accession': 'SRA_Accession'}, inplace=True)
# now concat
df_metadata_merged_v2 = pd.merge(df_metadata_merged_v1, df_metadata_sra_curator_v1, on='SRA_Accession', how='inner')
```

```python
df_metadata_merged_v2.to_csv('./output_qc/interim/sra_growth_metadata_ctherm_curated_p1.tsv', sep='\t') #first curation with curated file + SRA curation
```
