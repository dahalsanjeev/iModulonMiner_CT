### SIMULATION for all -- EH chemostat

```python
data_dir = path.join('..', 'ctherm_nf','processed_data')
interim_dir = path.join('..', 'ctherm_nf', 'interim')
external_dir = path.join('..', 'ctherm_nf', 'external')
```

```python
# for these experiments, we will first extract data for all imodulons
proj_of_interest = 'eh_chemostat_phenotype'

ethanol_col = 'Ethanol_g_L'
cellobiose_col = 'Cellobiose_feed_g_L'
acetate_col = 'Acetate_g_L'
lactate_col = 'Lactate_g_L'
formate_col = 'Formate_g_L'
pyruvate_col = 'Pyruvate_g_L'
pellet_col = 'Pellet_C_gC_L'
biomass_col = 'Dilution_rate_h-1'

dict_mw = {'Ethanol_g_L': model.metabolites.etoh_c.formula_weight,
           'Acetate_g_L': model.metabolites.ac_c.formula_weight,
           'Lactate_g_L': model.metabolites.lac__L_c.formula_weight,
           'Formate_g_L': model.metabolites.for_c.formula_weight,
           'Pyruvate_g_L': model.metabolites.pyr_c.formula_weight,
           'Cellobiose_feed_g_L': model.metabolites.cellb_c.formula_weight
}

# phenotype data to add
df_phenotype = pd.read_excel(path.join(external_dir, 'required_information_ctherm_imodulons.xlsx'), sheet_name=proj_of_interest)
df_phenotype = df_phenotype.loc[range(4,20)]
df_phenotype.columns = df_phenotype.loc[4].to_list()
df_phenotype.drop(4, inplace=True)
interested_strains_ciii_iv = ['LL376', 'LL1004', 'LL1036', 'LL374', 'LL1161', 'LL1042', 'LL1041', 'LL1043'] #EH_chemostat_CIII-IV
interested_strains_ci_ii = ['LL372', 'LL1004', 'LL345', 'LL1160', 'LL372', 'LL1044', 'LL1011', 'LL375'] #EH_chemostat_CI-II

# df_phenotype = df_phenotype.loc[df_phenotype.Strain.isin(interested_strains_ci_ii)]
# df_phenotype = df_phenotype.loc[df_phenotype.Strain.isin(interested_strains_ciii_iv)]

df_phenotype.set_index('Strain', inplace=True)
dict_pheno = df_phenotype[[ethanol_col, cellobiose_col, acetate_col, lactate_col, formate_col, pyruvate_col, pellet_col, biomass_col]].to_dict(orient='index')
```

### simulation for all the strains

```python
dict_strain_gene_ko = {'LL1004': [],
                        'LL345': ['Clo1313_2927'], #do not delete this gene, becomes infeasible
                        'LL1041': ['Clo1313_2927', 'Clo1313_1185'],
                        'LL1036': ['Clo1313_2927', 'Clo1313_1160'],
                        'LL1042': ['Clo1313_2927', 'Clo1313_1185', 'Clo1313_1160'],
                        'LL1011': ['Clo1313_2927', 'Clo1313_1185', 'Clo1313_1160'],
                        'LL1043': ['Clo1313_1185', 'Clo1313_1160', 'Clo1313_1185'], #hpt is added again
                        'LL376': ['Clo1313_2927', 'Clo1313_1409'], # spo0A: Clo1313_1409
                        'LL373': ['Clo1313_2927', 'Clo1313_1409', 'Clo1313_1185'], # spo0A: Clo1313_1409
                        'LL372': ['Clo1313_2927', 'Clo1313_1409', 'Clo1313_1160'], # spo0A: Clo1313_1409
                        'LL1044': ['Clo1313_1409', 'Clo1313_1160', 'Clo1313_1185'], # spo0A: Clo1313_1409, hpt is added again
                        'LL374': ['Clo1313_1409', 'Clo1313_1160', 'Clo1313_1185'], # spo0A: Clo1313_1409, hpt is added again
                        'LL375': ['Clo1313_1409', 'Clo1313_1160', 'Clo1313_1185'], # spo0A: Clo1313_1409, hpt is added again
                        'LL1160': ['Clo1313_2927'], # adhE is added again
                        'LL1161': ['Clo1313_2927'], # adhE is added again as mutated adhE_ins_D494G
}
```

```python
from cobra.flux_analysis import flux_variability_analysis as fva
```

```python
# universal
F_C =  60.055/114.12  # fraction of dry weight that is carbon (typical bacterial value) -- C5HO2N from Xiong et al., 2016 (10.1073/pnas.1605482113)

df_sol_strains = pd.DataFrame(columns=['sol_status', 'exp_etoh_conc', 'exp_growth_rate', 'pred_growth_rate', 'etoh_yield', 'pyr_yield', 'SO4_flux'])
dict_sol_ll1160_ll1004_ll1011 = dict() #to store solutions for LL1160, LL1004, and LL1011

for strain, sub_prod in dict_pheno.items():
    # global setting for strain
    dict_strain = dict_pheno[strain]
    D = dict_strain[biomass_col] #dilution rate
    X = dict_strain[pellet_col]/F_C #gDW/L

    dict_v_calc = dict() #computed fluxes from experimental data
    # now substrate, products
    for i in [ethanol_col, acetate_col, lactate_col, formate_col, pyruvate_col, cellobiose_col]:
        MW = dict_mw[i]
        C = dict_strain[i]
        dict_v_calc[i] = ((C*D) / (X*MW)) * 1000
    
    print(strain, D, dict_v_calc)
    
    # simulation for strains
    for r in model.reactions:
        r.bounds = dict_bounds[r.id]

    # biomass
    model.reactions.get_by_id('BIOMASS').bounds = D*0.95, D*1.05

    # set the uptake rates
    model.reactions.get_by_id('EXCH_cellb_e').bounds = -dict_v_calc[cellobiose_col]*1.1, -dict_v_calc[cellobiose_col]*0.9

    # set the secretion rates
    model.reactions.get_by_id('EXCH_ac_e').bounds = dict_v_calc[acetate_col]*0.9, dict_v_calc[acetate_col]*1.1 #acetate
    model.reactions.get_by_id('EXCH_lac__L_e').bounds = dict_v_calc[lactate_col]*0.9, dict_v_calc[lactate_col]*1.1 #lactate
    model.reactions.get_by_id('EXCH_for_e').bounds = dict_v_calc[formate_col]*0.9, dict_v_calc[formate_col]*1.1 #formate
    model.reactions.get_by_id('EXCH_pyr_e').bounds = dict_v_calc[pyruvate_col]*0.9, dict_v_calc[pyruvate_col]*1.1 #pyruvate
    model.reactions.get_by_id('EXCH_etoh_e').bounds = dict_v_calc[ethanol_col]*0.95, dict_v_calc[ethanol_col]*1.05 #ethanol

    # gene knockouts
    for gene in dict_strain_gene_ko[strain]:
        if gene=='Clo1313_2927': #don't delete the reactions associated with this
            continue
        try:
            gene = model.genes.get_by_id(gene)
        except KeyError:
            continue

        for rxn in gene.reactions:
            # print(rxn, sol_ref[rxn.id])
            if rxn.id in ['LDH_L', 'PTAr']:
                continue
            else:
                rxn.bounds = 0., 0.

    #FBA
    try:
        sol = pfba(model, fraction_of_optimum=0.9, objective={model.reactions.BIOMASS:1})
        df_sol_strains.loc[strain] = sol.status, dict_strain[ethanol_col], D, sol['BIOMASS'], np.abs(sol['EXCH_etoh_e']/sol['EXCH_cellb_e']), np.abs(sol['EXCH_pyr_e']/sol['EXCH_cellb_e']), sol['EXCH_so4_e']
    except cobra.exceptions.Infeasible:
        print(strain, 'Infeasible')
        df_sol_strains.loc[strain] = np.nan, dict_strain[ethanol_col], D, np.nan, np.nan, np.nan, np.nan

    if strain in ['LL1160', 'LL1004', 'LL1011']:
        dict_sol_ll1160_ll1004_ll1011[strain] = sol

# sort this
df_sol_strains = df_sol_strains.loc[df_sol_strains.sol_status=='optimal']
df_sol_strains.sort_values('etoh_yield', inplace=True)

# SO4 normalized by predicted growth rate
df_sol_strains['SO4_flux'] = df_sol_strains['SO4_flux'].div(df_sol_strains['pred_growth_rate'])

# save as another DF
df_sol_strains_eh = df_sol_strains.copy()
# to color code projects
df_sol_strains_eh['project'] = 'EH'
```

```python
# save dict_sol as a dataframe
df_sol_ll1160_ll1004_ll1011 = pd.DataFrame()

for k,v in dict_sol_ll1160_ll1004_ll1011.items():
    df_sol = v.to_frame()
    df_status = v.status
    df_obj = v.objective_value

    df_sol_ll1160_ll1004_ll1011[k] = df_sol['fluxes']
    df_sol_ll1160_ll1004_ll1011.loc['status', k] = df_status
    df_sol_ll1160_ll1004_ll1011.loc['obj_val', k] = df_obj


# save it as a DF
df_sol_ll1160_ll1004_ll1011.to_csv('./results_characterize_imodulons/sulfur_ethanol/sim_data_ll1160_ll1004_ll1011_solution_gsm.tsv', sep='\t')
```

### simulation for all the strains

```python
# universal
for strain, sub_prod in dict_pheno.items():
    if strain!='del_hpt_for0':
        continue
    # global setting for strain
    dict_strain = dict_pheno[strain]
    D = dict_strain[biomass_col] #dilution rate
    X = 1.04 * 0.27 #(1 OD600 = 1.04 gDW/L) from Xiong et al., 2016 (DOI: 10.1073/pnas.1605482113)

    dict_v_calc = dict() #computed fluxes from experimental data
    
    # now substrate, products
    for i in [ethanol_col, formate_col, acetate_col, lactate_col, cellobiose_col]:
        MW = dict_mw[i]
        C = dict_strain[i]
        dict_v_calc[i] = ((C*D) / (X*MW)) * 1000
    
    print(strain, dict_v_calc)
    
    # simulation for strains
    for r in model.reactions:
        r.bounds = dict_bounds[r.id]

    # media
    model.reactions.get_by_id('EXCH_cellb_e').bounds = -dict_v_calc[cellobiose_col]*1.1, -dict_v_calc[cellobiose_col]*0.9

    # biomass
    model.reactions.get_by_id('BIOMASS').bounds = D*0.95, D*1.05

    # set the secretion rates
    model.reactions.get_by_id('EXCH_ac_e').bounds = dict_v_calc[acetate_col]*0.9, dict_v_calc[acetate_col]*1.1 #acetate
    model.reactions.get_by_id('EXCH_lac__L_e').bounds = dict_v_calc[lactate_col]*0.9, dict_v_calc[lactate_col]*1.1 #lactate
    model.reactions.get_by_id('EXCH_for_e').bounds = dict_v_calc[formate_col]*0.9, dict_v_calc[formate_col]*1.1 #formate
    model.reactions.get_by_id('EXCH_etoh_e').bounds = dict_v_calc[ethanol_col]*0.95, dict_v_calc[ethanol_col]*1.05 #ethanol

    #FBA
    try:
        sol_ref = pfba(model, fraction_of_optimum=0.9, objective={model.reactions.BIOMASS:1})
    except cobra.exceptions.Infeasible:
        print('Infeasible solution')


# find solutions to reactions that will be knocked out
dict_sol_ref = dict()
for gene in ['Clo1313_2927', 'Clo1313_1160', 'Clo1313_1185', 'Clo1313_1186', 'Clo1313_1717', 'Clo1313_1571']:
    for rxn in model.genes.get_by_id(gene).reactions:
        dict_sol_ref[rxn.id] = sol_ref[rxn.id]
```

```python
# universal
df_sol_strains = pd.DataFrame(columns=['sol_status', 'exp_etoh_conc', 'exp_growth_rate', 'pred_growth_rate', 'etoh_yield', 'pyr_yield', 'for_flux', 'SO4_flux'])


LB, UB = 0., 0. #set both to zero to knock the reactions out

for strain, sub_prod in dict_pheno.items():
    # global setting for strain
    dict_strain = dict_pheno[strain]
    D = dict_strain[biomass_col] #dilution rate
    X = 1.04 * 0.27 #(1 OD600 = 1.04 gDW/L) from Xiong et al., 2016 (DOI: 10.1073/pnas.1605482113)

    dict_v_calc = dict() #computed fluxes from experimental data
    
    # now substrate, products
    for i in [ethanol_col, formate_col, acetate_col, lactate_col, cellobiose_col]:
        MW = dict_mw[i]
        C = dict_strain[i]
        dict_v_calc[i] = ((C*D) / (X*MW)) * 1000
    
    print(strain, D, dict_v_calc)
    
    # simulation for strains
    for r in model.reactions:
        r.bounds = dict_bounds[r.id]

    # media
    model.reactions.get_by_id('EXCH_cellb_e').bounds = -dict_v_calc[cellobiose_col]*1.1, -dict_v_calc[cellobiose_col]*0.9

    # biomass
    model.reactions.get_by_id('BIOMASS').bounds = D*0.95, D*1.05

    # set the secretion rates
    model.reactions.get_by_id('EXCH_ac_e').bounds = dict_v_calc[acetate_col]*0.9, dict_v_calc[acetate_col]*1.1 #acetate
    model.reactions.get_by_id('EXCH_lac__L_e').bounds = dict_v_calc[lactate_col]*0.9, dict_v_calc[lactate_col]*1.1 #lactate
    try:
        model.reactions.get_by_id('EXCH_for_e').bounds = dict_v_calc[formate_col]*0.9, dict_v_calc[formate_col]*1.1 #formate
    except ValueError:
        model.reactions.get_by_id('EXCH_for_e').bounds = dict_v_calc[formate_col]*1.1, 0. #formate
    
    model.reactions.get_by_id('EXCH_etoh_e').bounds = dict_v_calc[ethanol_col]*0.95, dict_v_calc[ethanol_col]*1.05 #ethanol

    # # constrain reactions for deletion
    if strain=='del_hpt_for0':
        # Dont delete 'Clo1313_2927'
        genes_to_delete = []
    else:
        genes_to_delete = ['Clo1313_1160', 'Clo1313_1185', 'Clo1313_1186', 'Clo1313_1717', 'Clo1313_1571']

    for gene in genes_to_delete:
        for rxn in model.genes.get_by_id(gene).reactions:
            if dict_sol_ref[rxn.id] < 0:
                rxn.bounds = UB * dict_sol_ref[rxn.id], LB * dict_sol_ref[rxn.id]
            else:
                rxn.bounds = LB * dict_sol_ref[rxn.id], UB * dict_sol_ref[rxn.id]

    #FBA
    # print(f'-------------{strain}-------------')
    try:
        sol = pfba(model, fraction_of_optimum=0.9, objective={model.reactions.BIOMASS:1})

        # for rxn in dict_sol_ref.keys():
        #     print(rxn, sol[rxn])

        df_sol_strains.loc[strain] = sol.status, dict_strain[ethanol_col], D, sol['BIOMASS'], np.abs(sol['EXCH_etoh_e']/sol['EXCH_cellb_e']), np.abs(sol['EXCH_pyr_e']/sol['EXCH_cellb_e']), sol['EXCH_for_e'], sol['EXCH_so4_e']
    except cobra.exceptions.Infeasible:
        df_sol_strains.loc[strain] = np.nan, dict_strain[ethanol_col], D, np.nan, np.nan, np.nan, np.nan, np.nan

# sort this
df_sol_strains = df_sol_strains.loc[df_sol_strains.sol_status=='optimal']
df_sol_strains.sort_values('etoh_yield', inplace=True)

# SO4 normalized by predicted growth rate
df_sol_strains['SO4_flux'] = df_sol_strains['SO4_flux'].div(df_sol_strains['pred_growth_rate'])

# make sure to replace labels
df_sol_strains['strain'] = ['del_hpt:for0', 'AG601:for2', 'AG553:for2', 'AG601:for0']
df_sol_strains.set_index('strain', inplace=True)

# save as another dataframe
df_sol_strains_evo = df_sol_strains.copy()
# for color coding projects
df_sol_strains_evo['project'] = 'evo'
```

```python
df_strain_flx = pd.DataFrame()

for strain in df_sol_ll1160_ll1004_ll1011.columns.to_list():    
    if strain=='LL1160':
        continue

    if df_sol_ll1160_ll1004_ll1011.loc['status', strain]!='optimal':
        raise ValueError(f'Solution for {strain} is not optimal.')

    for rxn in all_rxn:
        df_strain_flx.loc[strain, rxn[1]] = df_sol_ll1160_ll1004_ll1011.loc[rxn[1], strain]
    
    # # flux-sum for met of interest
    flx_sum = np.sum([np.abs(k[0]*eval(df_sol_ll1160_ll1004_ll1011.loc[k[1], strain])) for k in rxn_of_interest]) * 0.5
    df_strain_flx.loc[strain, f'flx_sum_{met_of_interest}'] = flx_sum
```
